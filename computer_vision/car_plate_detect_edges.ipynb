```python
import cv2
import matplotlib.pyplot as plt
import easyocr
```

```python
image = cv2.imread("data/car_plate.jpg")
img = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)
plt.imshow(image)
```

```python
gray = cv2.cvtColor(image , cv2.COLOR_RGB2GRAY)
blur = cv2.GaussianBlur(gray, (5 , 5), 0)

result = cv2.Canny(blur, 150, 200)
plt.imshow(result)
```

```python
reader = easyocr.Reader(['en'], gpu=True)

results = reader.readtext(result)

if len(results) == 0:
    print("No text was detected on the license plate.")
else:
    for (bbox, text, prob) in results:
        clean_text = "".join(e for e in text if e.isalnum() or e.isspace()).strip()
        
        print(f"Detected Plate Number: {clean_text}")
        print(f"Confidence Score: {prob:.4f}")
```
