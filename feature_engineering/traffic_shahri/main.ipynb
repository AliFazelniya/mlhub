```python
import numpy as np
import pandas as pd 
from khayyam import JalaliDatetime
from datetime import timedelta
import holidays
```

```python
df = pd.read_csv('data/traffic.csv', parse_dates=['DateTime'])
df.head()
```

```python
from datetime import datetime as dt
miladi_date = dt(1998,10,29)
print(f'Miladi date = {miladi_date}')

from khayyam import JalaliDatetime
shamsi_date = JalaliDatetime(miladi_date)
print('Sahmsi date = ', shamsi_date)
```

```python
df['JalaliDateTime'] = df["DateTime"].apply(JalaliDatetime)
df
```

```python
def cal_hour(hour):
    time = hour.strftime('%H:%M:%S')
    if '00:00:00' <= time <= '05:59:59':
        return 0

    elif '06:00:00' <= time <= '11:59:59':
        return 1
    
    elif '12:00:00' <= time <= '14:59:59':
        return 2
    elif '15:00:00' <= time <= '17:59:59':
        return 3
    elif '18:00:00' <= time <= '21:59:59':
        return 4
    elif '22:00:00' <= time <= '23:59:59':
        return 5


df['hour'] = df["JalaliDateTime"].apply(cal_hour)
df.head()
```

```python
iran_holidays = holidays.IR()
df['IsHoliday'] = df['JalaliDateTime'].apply(lambda d: 1 if d in iran_holidays else 0)
df.head()
```

```python
df['IsCold'] = df['JalaliDateTime'].apply(lambda d: 1 if d.month >= 7 else 0)
df.head()
```

```python
df = pd.concat([df, pd.get_dummies(df['Junction'], prefix='Junc')],axis=1)
df.head()
```
