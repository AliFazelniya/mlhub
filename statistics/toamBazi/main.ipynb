```python
import numpy as np
import pandas as pd
from scipy.stats import multinomial
```

```python
df = pd.read_csv("data/scores.csv")
df.head()
```

```python
df["avg"] = (df["Math"] + df["Physics"] + df["Literature"]) / 3
df.head()
```

```python
def score_changer(score):
    if score >= 70:
        return "A"
    elif score >= 50:
        return "B"
    else:
        return "C"
```

```python
df['FinalScore'] = df['avg'].apply(lambda x : score_changer(x))
df.head()
```

```python
joint_df = df.groupby(['Status','FinalScore']).size().unstack()/len(df)
joint_df
```

```python
score_marginal_distribution = joint_df.sum(axis=0)
```

```python
status_marginal_distribution = joint_df.sum(axis=1)
```

```python
p_A_given_book = joint_df.loc['bookworms', 'A'] / status_marginal_distribution['bookworms']
p_A_given_gamer = joint_df.loc['gamers', 'A'] / status_marginal_distribution['gamers']
nerds_to_gamers = round(abs((p_A_given_book / p_A_given_gamer) - 1) * 100 , 2)
nerds_to_gamers
```
