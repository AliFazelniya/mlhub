```python
import numpy as np
import pandas as pd
```

```python
df = pd.read_csv("data/students_performance.csv")
df.head()
```

```python
subjects = ["math score", "reading score", "writing score"]

mean_val = {}
median_val = {} 
mode_val = {}

for subject in subjects:

    mean_val[subject] = round(df[subject].mean(), 2)
    
    median_val[subject] = round(df[subject].median(), 2)
    
    mode_result = df[subject].mode()
    if len(mode_result) > 0:
        mode_val[subject] = round(float(mode_result.iloc[0]), 2)
    else:
        mode_val[subject] = float('nan')
```

```python
variance_val = {}
std_val = {}
range_val = {}

for subject in subjects:

    variance_val[subject] = round(df[subject].var(ddof=0), 2)
    
    std_val[subject] = round(df[subject].std(ddof=0), 2)
    
    range_val[subject] = round(df[subject].max() - df[subject].min(), 2)
```

```python
df["total score"] = df[['math score', 'reading score', 'writing score']].mean(axis=1)
df.head()
```

```python
Q1 = df['total score'].quantile(0.25)
Q3 = df['total score'].quantile(0.75)
IQR = Q3 - Q1
```

```python
less_than_Q1 = round(len(df[df['total score'] < Q1]) / len(df) * 100 , 2)
between_Q1_Q3 = round(len(df[(df['total score'] >= Q1) & (df['total score'] <= Q3)]) / len(df) * 100 , 2)
more_than_Q3 = round(len(df[df['total score'] > Q3]) / len(df) * 100 , 2)
```
