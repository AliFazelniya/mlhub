```python
import numpy as np
import pandas as pd
import plotly.express as px
```

```python
df = pd.read_csv("data/students_performance_updated.csv")
df.head()
```

```python
fig = px.box(df, y=["math score", "reading score", "writing score"], title="نمودار جعبه‌ای بررسی پراکندگی نمرات",
             labels={"value": "نمره", "variable": "درس‌ها"},
             color_discrete_sequence=["green"])
fig.show()
```

```python
fig = px.histogram(df, x="math score", title="هیستوگرام توزیع نمرات ریاضی",
                   labels={"value": "نمره", "variable": "ریاضی"},
                   color_discrete_sequence=["light blue"])

fig.show()
```
