```python
df = pd.read_csv('data/unidro_data.csv')
df.head()
```

```python
corr_matrix = df.corr()
```

```python
fig1, ax1 = plt.subplots(figsize=(12, 10))
sns.heatmap(corr_matrix, cmap='viridis', ax=ax1)
ax1.set_title("correlation heatmap")
```

```python
my_mask = np.triu(np.ones_like(corr_matrix, dtype=bool))
fig2, ax2 = plt.subplots(figsize=(12, 10))
sns.heatmap(corr_matrix, cmap='jet', mask=my_mask, ax=ax2)
ax2.set_title("correlation heatmap lower triangular")
```
