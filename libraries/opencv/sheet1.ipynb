```python
import cv2
import matplotlib.pyplot as plt
import numpy as np
```

```python
image = cv2.imread("data/images/coffee.jpg")
plt.imshow(image)
plt.show()
```

```python
print(f"image.shape : {image.shape}")
print(f"image dtype : {image.dtype}")
```

```python
image_crop = image[100:900, 200:800]
plt.imshow(image_crop)
plt.show()
```

```python
b, g, r = cv2.split(image)
image_merged = cv2.merge((r, g, b))
plt.imshow(image_merged)
plt.show()
```

```python
image_resized = cv2.resize(image, None, fx = 0.25, fy = 0.25)
plt.imshow(image_resized)
plt.show()
```

```python
image_drink = cv2.imread("data/images/drink.jpg")
plt.imshow(image_drink)
plt.show()
```

```python
image_drink = cv2.cvtColor(image_drink, cv2.COLOR_BGR2RGB)
plt.imshow(image_drink)
plt.show()
```

```python
image_drink_line = cv2.line(image_drink, (10, 300), (100, 1000), (0, 0, 0), 2)
plt.imshow(image_drink_line)
plt.show()
```

```python
image_drink_rectangle = cv2.rectangle(image_drink, (100,100), (400, 400), (0, 0, 255), 2)
plt.imshow(image_drink_rectangle)
plt.show()
```

```python
image_circle = cv2.circle(image_drink, (250,250), 150, (0, 255, 0), 2)
plt.imshow(image_circle)
plt.show()
```

```python
pts = np.array([[10,10], [60, 10], [70, 30], [20, 30]])
image_drink_polylines = cv2.polylines(image_drink, [pts], True, (0, 0, 0), 3)
plt.imshow(image_drink_polylines)
plt.show()
```

```python
image_lake = cv2.imread('data/images/lake.jpg')
plt.imshow(image_lake)
plt.show()
```

```python
image_hsv = cv2.cvtColor(image_lake, cv2.COLOR_BGR2HSV)

mark = cv2.inRange(image_hsv, np.array([15, 0, 0], dtype = np.uint8), np.array([30, 255, 255], dtype=np.uint8))
image_hsv[mark > 0] -= np.array([15, 0, 0], dtype = np.uint8)

plt.imshow(image_hsv)
plt.show()
```

```python
image_lake_bgr = cv2.cvtColor(image_hsv, cv2.COLOR_HSV2BGR)
plt.imshow(image_lake_bgr)
plt.show()
```

```python
image_medic = cv2.imread("data/images/medic.jpg")
plt.imshow(image_medic)
plt.show()
```

```python
h, w, ch = image_medic.shape
pts = np.array([[60,50], [290, 60], [342, 408], [15, 415]], dtype = np.float32)
pts2 = np.array([[0, 0], [w, 0], [w, h], [0, h]], dtype = np.float32)
image_medic_polylines = cv2.polylines(image_medic, [np.int32(pts)], True, (0, 255, 0), 2)

plt.imshow(image_medic_polylines)
plt.show()
```

```python
m = cv2.getPerspectiveTransform(pts, pts2)
image_warp = cv2.warpPerspective(image_medic, m, (w, h))
plt.imshow(image_warp)
plt.show()
```

```python
m = cv2.getRotationMatrix2D((w//2, h//2), 30, 1)
image_warp_affine = cv2.warpAffine(image_medic, m, (w, h))
plt.imshow(image_warp_affine)
plt.show()
```

```python
image_splash = cv2.imread("data/images/unsplash.jpg")
plt.imshow(image_splash)
plt.show()
```

```python
k = np.array([[-1, 0, 1], [-2, 0, 2], [-1, 0 ,1]])
image_filter = cv2.filter2D(image_splash, -1, k)
plt.imshow(image_filter)
plt.show()
```

```python
image_blur = cv2.blur(image_splash, (5, 5))
plt.imshow(image_blur)
plt.show()
```

```python
image_gaussian = cv2.GaussianBlur(image_splash, (5, 5), 0)
plt.imshow(image_gaussian)
plt.show()
```

```python
image_median = cv2.medianBlur(image_splash, 7)
plt.imshow(image_median)
plt.show()
```

```python
image_sobel = cv2.Sobel(image_splash, cv2.CV_64F, 1, 0, (3, 3))
plt.imshow(image_sobel)
plt.show()
```

```python
image_sobel = cv2.Sobel(image_splash, cv2.CV_64F, 0, 1, (3, 3))
plt.imshow(image_sobel)
plt.show()
```

```python
image_car_plate = cv2.imread("data/images/car_plate.jpg")
image_car_plate = cv2.cvtColor(image_car_plate, cv2.COLOR_BGR2RGB)
plt.imshow(image_car_plate)
plt.show()
```

```python
result = cv2.Canny(image_car_plate, 150, 200)
plt.imshow(result)
plt.show()
```

```python
blur = cv2.GaussianBlur(image_car_plate, (5 , 5), 0)

result = cv2.Canny(blur, 150, 200)
plt.imshow(result)
```

```python
gray = cv2.cvtColor(image_car_plate , cv2.COLOR_RGB2GRAY)
blur = cv2.GaussianBlur(gray, (5 , 5), 0)

result = cv2.Canny(blur, 150, 200)
plt.imshow(result)
```

```python
image_drink_cake = cv2.imread("data/images/drink_cake.jpg")
plt.imshow(image_drink_cake)
plt.show()
```

```python
hist = cv2.calcHist([image_drink_cake], [0], None, [256], [0,256])
plt.plot(hist)
plt.show()
```

```python
gray = cv2.cvtColor(image_drink_cake, cv2.COLOR_BGR2GRAY)
image2 = cv2.equalizeHist(gray)

hist2 = cv2.calcHist([image2], [0], None, [256], [0,256])
plt.plot(hist2)
plt.show()
```

```python
gray = cv2.cvtColor(image_drink_cake, cv2.COLOR_BGR2GRAY)

clahe = cv2.createCLAHE(clipLimit=2.0, tileGridSize=(8,8))
image2 = clahe.apply(gray)

plt.imshow(image2, cmap='gray')
plt.axis('off')
plt.show()
```
