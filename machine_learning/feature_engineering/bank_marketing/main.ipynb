```python
import pandas as pd
```

```python
df = pd.read_csv('data/bank-additional-full.csv', na_values='unknown')
df.head()
```

```python
categorical_cols = ["month", "education", "day_of_week", "poutcome", "loan", "housing", "default", "job", "marital" , "is_telephone_contact"]
df[categorical_cols] = df[categorical_cols].apply(
    lambda s: s.fillna(s.mode(dropna=True).iloc[0] if not s.mode(dropna=True).empty else s)
)
df.head()
df.isna().sum()
```

```python
continuous_cols = ["duration", "age", "pdays", "previous", "campaign"]
binary_cols = ["loan", "housing", "default" , "is_telephone_contact"]
nominal_cols = ["marital" , "job"]
ordinal_cols = ["education", "month", "day_of_week", "poutcome"]
```

```python
mapping = {'no': 0,
           'yes' : 1}

binary_df = df[binary_cols]

for i in binary_cols:
    binary_df[i] = binary_df[i].map(mapping)

binary_df.head()
```

```python
nominal_df = df[nominal_cols]
nominal_df = pd.get_dummies(
    nominal_df,
    prefix=nominal_cols,
    prefix_sep='_', 
    dtype=int  
)

nominal_df.head()
```

```python
mapping1 = {
    "illiterate" : 0,
    "basic.4y" : 1,
    "basic.6y" : 2,
    "basic.9y" : 3,
    "high.school" : 4,
    "university.degree" : 5,
    
}

mapping2 ={
    "mon" : 0,
    "tue" : 1,
    "wed" : 2,
    "thu" : 3,
    "fri" : 4,
    "sat" : 5,
    "sun" : 6
}

mapping3 = {
    "jan" : 0,
    "feb" : 1,
    "mar" : 2,
    "apr" : 3,
    "may" : 4,
    "jun" : 5,
    "jul" : 6,
    "aug" : 7,
    "sep" : 8,
    "oct" : 9,
    "nov" : 10,
    "dec" : 11 
}

mapping4 = {
    "failure" : -1,
    "nonexistent" : 0,
    "success" : 1
}

ordinal_df = df[ordinal_cols]
ordinal_df["education"] = ordinal_df["education"].map(mapping1)
ordinal_df["month"] = ordinal_df["month"].map(mapping3)
ordinal_df["day_of_week"] = ordinal_df["day_of_week"].map(mapping2)
ordinal_df["poutcome"] = ordinal_df["poutcome"].map(mapping4)

ordinal_df.head()
```
