    تکنیک‌هایی که استفاده می‌کنید به شکل مستقیم مورد ارزیابی توسط سامانه داوری قرار <b>نمی‌گیرند.</b> بلکه همه آن‌ها در دقت مدل شما تاثیر خواهند گذاشت؛ بنابراین هر چه پیش‌پردازش/مهندسی ویژگی بهتری انجام دهید تا دقت مدل بهبود پیدا کند، امتیاز بیشتری از این سوال کسب خواهید کرد.

```python
FEMALE_NAMES_FA = [
    'زهرا','فاطمه','مریم','سارا','نازنین','ندا','نیلوفر','شیما','شیرین',
    'لیلا','رویا','سپیده','آتنا','سمیرا','الناز','نگار','آزاده','مهسا',
    'نسرین','پریسا','یاسمن','یاسمین','مهناز','شادی','بهاره','فرناز',
    'گلناز','فریبا','مونا','منا','هانیه','هانیا','دنیا','تارا','نیکا',
    'رها','آنا','ژاله','آرزو','سوسن','الهه','معصومه','طاهره','زینب',
    'لنا','باران','ستاره','افسانه','صبا','غزل','نرجس','ریحانه','فائزه',
    'حدیث','سحر','ملیحه','ماندانا','مهتاب','مهلا','ترانه','پرستو',
    'نیوشا','پگاه','سیمین','کیمیا','آیلین','آیدا','روژان','هلیا','آوا',
    'خانم','بانو','دختر','ملیسا','رزا','ویانا','آریانا','ایلین','میترا',
    'پریا','ساغر','نادیا','نیلا','شادن','سوفیا','آناهیتا','درسا','تینا'
]

MALE_NAMES_FA = [
    'محمد','علی','حسین','رضا','مهدی','امیر','سعید','حمید','کاوه',
    'بهزاد','پویا','آرش','میلاد','پدرام','فرشاد','سجاد','مجتبی',
    'احمد','مصطفی','حسن','ابراهیم','اکبر','اصغر','عباس','جواد',
    'امیررضا','اميررضا','علیرضا','محمدرضا','محمدعلی','محمدامین',
    'سامان','شاهین','نیما','ایمان','کیان','آرمان','آرمین','دانیال',
    'بابک','فرهاد','کامران','کیانوش','داریوش','فرشید','ساسان',
    'عرفان','پرهام','سینا','امید','نوید','هادی','یاسر','ابوالفضل',
    'امیرحسین','حامد','وحید','شهاب','صادق','مرتضی','رامین','شاهرخ',
    'پارسا','کورش','آقا','جناب','پسر','سیاوش','کامیار','حمزه','یونس'
]

FEMALE_NAMES_EN = [
    'zahra','fateme','maryam','sara','nazanin','neda','shima','shirin',
    'leila','roya','sepideh','atena','samira','elnaz','negar','azadeh',
    'mahsa','nasrin','parisa','yasaman','yasmin','mahnaz','shadi','bahar',
    'farnaz','golnaz','fariba','mona','hanie','tara','nika','raha',
    'ghazal','saba','sahar','reyhane','hadis','baran','setare','niloufar',
    'kimia','ailin','aida','rozhan','helia','ava','ainaz','sofia','parya',
    'narsis','narges','nargess','melisa','roza','viana','ariana','mitra',
    'lady','girl','mama','mommy','mom','miss','wife','bride','queen','princess',
    'dokhtar','khanom','banu'
]

MALE_NAMES_EN = [
    'mohammad','ali','hossein','reza','mehdi','amir','saeed','hamid',
    'kaveh','behzad','pouya','arash','milad','pedram','farshad','sajjad',
    'mostafa','hasan','ebrahim','saman','shahin','nima','iman','kian',
    'arman','armin','danial','babak','farhad','kamran','dariush','farshid',
    'sasan','erfan','parham','sina','omid','navid','hadi','yaser',
    'amirhossein','alireza','mohammadreza','hamed','vahid','morteza','ramin',
    'shahab','sadegh','parsa','kourosh','siavash','kamyar','hamzeh',
    'boy','mr','sir','king','prince','father','dad','daddy','bro','agha','pesar'
]

FEMALE_EMOJIS = ['👩','💅','👗','👠','👒','💄','👑','🌸','🌺','🌷',
                 '💋','👰','🤰','🧕','💃','🌹','🎀','💝','💖','💗',
                 '👸','🫶','💞','🌼','🌻','🦋','🪷','🫦','🩷','🎀']
MALE_EMOJIS   = ['👨','🤴','💪','⚽','🏋','🏊','🚴','🎮','🏈',
                 '🤜','👊','🧔','🤵','🎯','🤺','🏇','🔧','⚙️','🛠','🏃']

def count_matches(text, word_list):
    t = str(text).lower()
    return sum(1 for w in word_list if w in t)


def engineer_features(df):

    feats = pd.DataFrame(index=df.index)

    fullname  = df['fullname'].fillna('').str.strip()
    username  = df['username'].fillna('').str.strip()
    biography = df['biography'].fillna('').str.strip()

    feats['age']             = df['age'].fillna(df['age'].median())
    feats['follower_count']  = df['follower_count'].fillna(0)
    feats['following_count'] = df['following_count'].fillna(0)
    feats['is_business']     = df['is_business'].fillna(0).astype(int)
    feats['is_verified']     = df['is_verified'].fillna(0).astype(int)
    feats['is_private']      = df['is_private'].fillna(0).astype(int)

    feats['follower_log']  = np.log1p(feats['follower_count'])
    feats['following_log'] = np.log1p(feats['following_count'])
    feats['follow_ratio']  = feats['follower_count'] / (feats['following_count'] + 1)
    feats['follow_diff']   = feats['follower_count'] - feats['following_count']
    feats['has_followers'] = (feats['follower_count'] > 0).astype(int)
    feats['high_follower'] = (feats['follower_count'] > 5000).astype(int)

    feats['name_female_fa'] = fullname.apply(lambda x: count_matches(x, FEMALE_NAMES_FA))
    feats['name_male_fa']   = fullname.apply(lambda x: count_matches(x, MALE_NAMES_FA))
    feats['name_female_en'] = fullname.apply(lambda x: count_matches(x, FEMALE_NAMES_EN))
    feats['name_male_en']   = fullname.apply(lambda x: count_matches(x, MALE_NAMES_EN))

    feats['name_signal'] = (
        feats['name_female_fa'] * 3 + feats['name_female_en'] -
        feats['name_male_fa']   * 3 - feats['name_male_en']
    )
    feats['name_has_lady']   = fullname.str.lower().str.contains(
        r'lady|girl|\bms\b|\bmrs\b|بانو|خانم', regex=True, na=False).astype(int)
    feats['name_has_mr']     = fullname.str.lower().str.contains(
        r'\bmr\b|\bsir\b|آقا|جناب', regex=True, na=False).astype(int)
    feats['name_has_mommy']  = fullname.str.lower().str.contains(
        r'mommy|mama|mom|مامان|مادر', regex=True, na=False).astype(int)

    feats['name_len']        = fullname.str.len()
    feats['name_word_count'] = fullname.str.split().str.len().fillna(0)

    feats['user_female_fa'] = username.apply(lambda x: count_matches(x, FEMALE_NAMES_FA))
    feats['user_male_fa']   = username.apply(lambda x: count_matches(x, MALE_NAMES_FA))
    feats['user_female_en'] = username.apply(lambda x: count_matches(x, FEMALE_NAMES_EN))
    feats['user_male_en']   = username.apply(lambda x: count_matches(x, MALE_NAMES_EN))
    feats['user_signal']    = (
        feats['user_female_fa'] * 3 + feats['user_female_en'] -
        feats['user_male_fa']   * 3 - feats['user_male_en']
    )

    feats['user_has_mr']     = username.str.lower().str.contains(
        r'^mr[_.]|[_.]mr[_.]|^mr$', regex=True, na=False).astype(int)
    feats['user_has_lady']   = username.str.lower().str.contains(
        r'lady|girl|\bms\b', regex=True, na=False).astype(int)
    feats['user_has_mommy']  = username.str.lower().str.contains(
        r'mommy|mama|mom', regex=True, na=False).astype(int)
    feats['user_len']        = username.str.len()
    feats['user_dot_count']  = username.str.count(r'\.').fillna(0)
    feats['user_underscore'] = username.str.count('_').fillna(0)
    feats['user_has_digits'] = username.str.contains(r'\d', na=False).astype(int)

    feats['bio_len']  = biography.str.len()
    feats['has_bio']  = (feats['bio_len'] > 0).astype(int)

    feats['bio_female_emoji'] = biography.apply(lambda x: count_matches(x, FEMALE_EMOJIS))
    feats['bio_male_emoji']   = biography.apply(lambda x: count_matches(x, MALE_EMOJIS))
    feats['bio_emoji_signal'] = feats['bio_female_emoji'] - feats['bio_male_emoji']

    feats['bio_female_name'] = biography.apply(lambda x: count_matches(x, FEMALE_NAMES_FA[:30] + FEMALE_NAMES_EN[:15]))
    feats['bio_male_name']   = biography.apply(lambda x: count_matches(x, MALE_NAMES_FA[:30] + MALE_NAMES_EN[:15]))

    bio_lower = biography.str.lower()
    feats['bio_wife']     = bio_lower.str.contains('wife|همسر|زوجه|عروس', regex=True, na=False).astype(int)
    feats['bio_mom']      = bio_lower.str.contains('mom|mama|mommy|مامان|مادر', regex=True, na=False).astype(int)
    feats['bio_husband']  = bio_lower.str.contains('husband|شوهر', regex=True, na=False).astype(int)
    feats['bio_dad']      = bio_lower.str.contains('dad|father|بابا|پدر', regex=True, na=False).astype(int)
    feats['bio_sport']    = bio_lower.str.contains('sport|ورزش|football|فوتبال|بدنساز', regex=True, na=False).astype(int)
    feats['bio_beauty']   = bio_lower.str.contains('beauty|زیبا|آرایش|مد|fashion', regex=True, na=False).astype(int)
    feats['bio_engineer'] = bio_lower.str.contains('engineer|مهندس', regex=True, na=False).astype(int)

    feats['total_female_score'] = (
        feats['name_female_fa'] * 4 +
        feats['user_female_fa'] * 3 +
        feats['name_female_en'] * 2 +
        feats['user_female_en'] * 2 +
        feats['bio_female_emoji'] +
        feats['bio_female_name'] +
        feats['name_has_lady'] * 3 +
        feats['user_has_lady'] * 3 +
        feats['bio_wife'] * 2 +
        feats['bio_mom'] * 2 +
        feats['name_has_mommy'] * 2 +
        feats['user_has_mommy'] * 2
    )
    feats['total_male_score'] = (
        feats['name_male_fa'] * 4 +
        feats['user_male_fa'] * 3 +
        feats['name_male_en'] * 2 +
        feats['user_male_en'] * 2 +
        feats['bio_male_emoji'] +
        feats['bio_male_name'] +
        feats['name_has_mr'] * 3 +
        feats['user_has_mr'] * 3 +
        feats['bio_husband'] * 2 +
        feats['bio_dad'] * 2 +
        feats['bio_sport']
    )
    feats['overall_signal'] = feats['total_female_score'] - feats['total_male_score']

    return feats.fillna(0)


X_all  = engineer_features(train)

X_test = engineer_features(test)

y_all = (train['gender'] == 'woman').astype(int)


tmp = X_all[['overall_signal']].copy()
tmp['gender'] = train['gender'].values

X_tr, X_val, y_tr, y_val = train_test_split(
    X_all, y_all, test_size=0.15, random_state=42, stratify=y_all
)
```

<h2 align=right style="line-height:200%;font-family:vazir;color:#0099cc">
<font face="vazirmatn" color="#0099cc">
مدل‌سازی
</font>
</h2>

<p dir=rtl style="direction: rtl; text-align: justify; line-height:200%; font-family:vazir; font-size:medium">
<font face="vazirmatn" size=3>
    حال که داده را پاکسازی کرده و احتمالا ویژگی‌هایی را به آن افزوده یا از آن حذف کرده‌اید، وقت آن است که مدلی آموزش دهید که بتواند متغیر هدف این مسئله را پیش‌بینی کند.
</font>
</p>

```python
rf = RandomForestClassifier(
    n_estimators=500,
    max_depth=None,
    min_samples_leaf=2,
    max_features='sqrt',
    class_weight='balanced',
    n_jobs=-1,
    random_state=42,
)

rf.fit(X_tr, y_tr)
rf_f1 = f1_score(y_val, rf.predict(X_val), average='weighted')
print(f'Random Forest Val F1: {rf_f1:.4f} ({rf_f1*100:.2f}%)')

print()

gb = GradientBoostingClassifier(
    n_estimators=300,
    learning_rate=0.05,
    max_depth=5,
    min_samples_leaf=10,
    subsample=0.8,
    random_state=42,
)

gb.fit(X_tr, y_tr)
gb_f1 = f1_score(y_val, gb.predict(X_val), average='weighted')
print(f'Gradient Boosting Val F1: {gb_f1:.4f} ({gb_f1*100:.2f}%)')
```

<h3 align=right style="line-height:200%;font-family:vazir;color:#0099cc">
<font face="vazirmatn" color="#0099cc">
معیار ارزیابی
</font>
</h3>

<p dir=rtl style="direction: rtl; text-align: justify; line-height:200%; font-family:vazir; font-size:medium">
<font face="vazirmatn" size=3>
    معیاری که برای ارزیابی عملکرد مدل انتخاب کرده‌ایم، <code>f1_score</code> نام دارد.
    <br>
    این معیار، سنجه ارزیابی کیفیت مدل شماست. به عبارت بهتر در سامانه داوری هم از همین معیار برای نمره‌دهی استفاده شده است.
    <br>
    پیشنهاد می‌شود با توجه به این معیار، عملکرد مدل خود را بر روی دادگان آموزش یا اعتبارسنجی ارزیابی کنید.
</font>
</p>

```python
if gb_f1 >= rf_f1:
    best_val_model = gb
    best_val_f1 = gb_f1
    model_name = 'Gradient Boosting'
else:
    best_val_model = rf
    best_val_f1 = rf_f1
    model_name = 'Random Forest'

print(f'Best model: {model_name}')
print(f'Validation Weighted F1: {best_val_f1:.4f} ({best_val_f1*100:.2f}%)')
print(f'Target: ≥ 75% | Status: {"✅ PASS" if best_val_f1 >= 0.75 else "❌ Below target"}')
print()

y_pred_val = best_val_model.predict(X_val)
label_map  = {0: 'man', 1: 'woman'}
print(classification_report(
    pd.Series(y_val.values).map(label_map),
    pd.Series(y_pred_val).map(label_map)
))
```
