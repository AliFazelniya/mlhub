```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
def sigmoid(x):
    return 1 / (1 + np.exp(-x))

def sigmoid_derivative(x):
    return sigmoid(x) * (1 - sigmoid(x))

x_data = np.linspace(-10, 10, 100)
y_sigmoid = sigmoid(x_data)
dy_data = sigmoid_derivative(x_data)

plt.plot(x_data, y_sigmoid, x_data, dy_data)
plt.title('Sigmoid Activation Function')
plt.legend(['sigmoid', 'sigmoid derivative'])
plt.grid()
plt.show()
```

```python
def tanh(x):
    return (np.exp(x) - np.exp(-x)) / (np.exp(x) + np.exp(-x))

def tanh_derivative(x):
    return 1 - tanh(x) ** 2


x_data = np.linspace(-6, 6, 100)
y_tanh = tanh(x_data)
dy_tanh = tanh_derivative(x_data)

plt.plot(x_data, y_tanh, x_data, dy_tanh)
plt.title('Hyperbolic Tangent Activation Function')
plt.legend(['tanh', 'tanh derivative'])
plt.grid()
plt.show()
```

```python
def relu(x):
    data = [max(0, i) for i in x]
    return np.array(data , dtype = float)

def relu_derivative(x):
    data = [1 if i > 0 else 0 for i in x]
    return np.array(data , dtype = float)

x_data = np.linspace(-10, 10, 100)
y_relu = relu(x_data)
dy_relu = relu_derivative(x_data)

plt.plot(x_data, y_relu, x_data, dy_relu)
plt.title('ReLU Activation Function')
plt.legend(['ReLU', 'ReLU derivative'])
plt.grid()
plt.show()
```

```python
def softmax(x):
    exp_x = np.exp(x - np.max(x))
    return exp_x / exp_x.sum(axis = 0, keepdims = True)

x_data = np.linspace(-10, 10, 100)
y_softmax = softmax(x_data)
plt.plot(x_data, y_softmax)
plt.title('Softmax Activation Function')
plt.legend(['softmax'])
plt.grid()
plt.show()
```
