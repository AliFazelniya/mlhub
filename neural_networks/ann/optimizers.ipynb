```python
import numpy as np
import matplotlib.pyplot as plt
from tensorflow.keras.optimizers import SGD , Adagrad, Adadelta, RMSprop, Adam
```

```python
def func(x):
    return (x ** 2) - (2 * x) - 3

def fprime(x):
    return 2 * x - 2

def plotFunc(x0):
    x = np.linspace(-5 , 7 , 100)
    plt.plot(x , func(x))
    plt.plot(x0, func(x0), 'ko')
    plt.xlabel('$x$')
    plt.ylabel('$f(x)$')
    plt.title('Objective Function')

def plotPath(xs , ys, x0):
    plotFunc(x0)
    plt.plot(xs , ys , linestyle = '--', marker = 'o' , color = '#F12F79')
    plt.plot(xs[-1], ys[-1], 'ko')

x0 = -4
def gradientdesent(func , fprime , x0, alpha, tol = 1e-5, max_iter = 1000):
    xk = x0
    fk = func(xk)
    pk = - fprime(xk)

    num_iter = 0
    curve_x = [xk]
    curve_y = [fk]

    while abs(pk) > tol and num_iter < max_iter:
        xk = xk + alpha * pk
        fk = func(xk)
        pk = - fprime(xk)

        num_iter += 1
        curve_x.append(xk)
        curve_y.append(fk)

    if num_iter == max_iter:
        print('gradient decent does not converge')
    else:
        print(f'Solution found:\n y = {fk:.4f}\n x + {xk:.4f}')

    return curve_x , curve_y

print("=============================== Scenario 1 ===============================")
xs , ys = gradientdesent(func, fprime , x0, alpha = 0.1)
plotPath(xs , ys, x0)
plt.show()

print("=============================== Scenario 2 ===============================")
xs , ys = gradientdesent(func, fprime , x0, alpha = 0.9)
plotPath(xs , ys, x0)
plt.show()

print("=============================== Scenario 3 ===============================")
xs , ys = gradientdesent(func, fprime , x0, alpha = 1e-4)
plotPath(xs , ys, x0)
plt.show()

print("=============================== Scenario 4 ===============================")
xs , ys = gradientdesent(func, fprime , x0, alpha = 1.01, max_iter = 8)
plotPath(xs , ys, x0)
plt.show()
```
