```python
def sigmoid(x, a):
    y = 1 / (1 + np.exp(-a*x))
    return y
```

```python
x = np.linspace(-20, 20, 100)
y = sigmoid(x , a = 1)
plt.plot(x , y)
```

```python
x = np.linspace(-20, 20, 100)
plt.plot(x , sigmoid(x , a = 0.1))
```

```python
plt.plot(x , sigmoid(x , a = 2))
```
