```python
import pandas as pd
from datetime import datetime
```

```python
df = pd.read_csv("data/job_posts.csv")
df.head()
```

```python
df = df.drop(df[(df['Salary'].isna()) & (df['Level'].isna()) & (df['Collaboration Type'].isna())].index)
```

```python
df['Salary'] = pd.to_numeric(df['Salary'].str.replace("MT", "").str.replace("<", "").str.replace(">", ""), errors='coerce')
df
```

```python
df['Salary'] = df['Salary'].fillna(df.groupby('Level')['Salary'].transform('mean'))
```

```python
start_of_corona = datetime.strptime("2020-02-19", "%Y-%m-%d")
```

```python
remote_growth = (df[(df['Offers Remote'] == True) & (pd.to_datetime(df['Publish Time']) >= start_of_corona)].shape[0] / df.shape[0]) - (df[(df['Offers Remote'] == True) & (pd.to_datetime(df['Publish Time']) < start_of_corona)].shape[0] / df.shape[0])
```

```python
discrete_time = pd.cut(pd.to_datetime(df['Publish Time']).dt.hour, bins=[0, 6, 12, 18, 24], labels=['dawn', 'morning', 'noon', 'night'], right=False).value_counts().sort_values(ascending=False)
```
