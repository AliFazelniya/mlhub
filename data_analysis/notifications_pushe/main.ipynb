```python
df = pd.read_csv("data/notifications_pushe.csv")
df
```

```python
df['date'] = pd.to_datetime(pd.to_datetime("2019-06-01") + pd.to_timedelta(df['interaction_time'].astype(int), unit='s'))
df
```

```python
tuple_appid_successrate = (int(df.groupby('application_id').apply(lambda group: (group['action'] == 'C').sum() / len(group), include_groups=False).idxmax()), df.groupby('application_id').apply(lambda group: (group['action'] == 'C').sum() / len(group), include_groups=False).max())
tuple_appid_successrate
```

```python
tuple_notifid_day = (df.query("action == 'C'").groupby('notification_id').size().div(df.groupby('notification_id').size()).idxmax(), df[df['notification_id'] == df.query("action == 'C'").groupby('notification_id').size().div(df.groupby('notification_id').size()).idxmax()].assign(weekday = pd.to_datetime(df['date']).dt.day_name()).query("action == 'C'").groupby('weekday').size().idxmax())
tuple_notifid_day
```

```python
tuple_day_rate = df.groupby(df['date'].dt.day).apply(lambda group: (group['action'] == 'C').sum() / len(group)).idxmax(), df.groupby(df['date'].dt.day).apply(lambda group: (group['action'] == 'C').sum() / len(group)).max()
tuple_day_rate
```
