```python
import pandas as pd

df = pd.read_csv('data/data.csv')
df.head()
```

```python
df['avg_sal'] = df.groupby('DESCRSHORT')['COMPRATE'].transform('mean')
df.head(3)
```

```python
above_avg = df[df['COMPRATE'] > df['avg_sal']]
above_avg.head(3)
```

```python
above_average_employees_df = above_avg[
    ['OBJECTID', 'DESCRSHORT', 'COMPRATE']
]
above_average_employees_df.head(3)
```

```python
above_average_employees_df = above_average_employees_df.sort_values('OBJECTID')
above_average_employees_df.head(3)
```

```python
df['HIREDATE'] = pd.to_datetime(df['HIREDATE_STRING'])
df.head(3)
```

```python
df['is_loyal'] = df['HIREDATE'].dt.year < 2000
df.head(3)
```

```python
dept_stats = df.groupby('DESCRSHORT').agg(
    total_employees=('OBJECTID', 'count'),
    loyal_employees=('is_loyal', 'sum')
).reset_index()
dept_stats.head(3)
```

```python
dept_stats = dept_stats[dept_stats['total_employees'] >= 10]
dept_stats.head(3)
```

```python
dept_stats['Loyal_Employee_Percentage'] = (
    dept_stats['loyal_employees'] / dept_stats['total_employees'] * 100
).round(2)
dept_stats.head(3)
```

```python
dept_stats = dept_stats[dept_stats['loyal_employees'] > 0]
dept_stats.head(3)
```

```python
loyal_percentage_df = dept_stats[
    ['DESCRSHORT', 'Loyal_Employee_Percentage']
].sort_values(by='Loyal_Employee_Percentage', ascending=False)

loyal_percentage_df.head(3)
```

```python
df['HIREDATE_STRING'] = pd.to_datetime(df['HIREDATE_STRING'], errors='coerce')
df['year'] = df['HIREDATE_STRING'].dt.year
df.head(3)
```

```python
df['quintile'] = pd.qcut(df['year'], q=5, labels=[5, 4, 3, 2, 1])
df.head(3)
```

```python
avg_salary = df.groupby('quintile')['COMPRATE'].mean().round(2)
avg_salary
```

```python
average_salary_by_quintile_df = avg_salary.reset_index(
    name='Average_Salary'
)
average_salary_by_quintile_df
```

```python
average_salary_by_quintile_df['quintile'] = average_salary_by_quintile_df['quintile'].astype(int)
average_salary_by_quintile_df
```

```python
average_salary_by_quintile_df = average_salary_by_quintile_df.rename(
    columns={'quintile': 'Tenure_Quintile'}
)
average_salary_by_quintile_df
```

```python
average_salary_by_quintile_df = average_salary_by_quintile_df.sort_values(
    'Tenure_Quintile',
    ascending=False
)

# Final result
average_salary_by_quintile_df
```
