```python
import numpy as np
import pandas as pd 
```

```python
df = pd.read_csv("data/migration_rate.csv")
```

```python
df = df.dropna(thresh=5)
df.info
```

```python
df.iloc[:, 1:] = df.iloc[:, 1:].apply(lambda row: row.fillna(row.mean()), axis=1)
```

```python
df.isna().sum()
```

```python
top_countries = df.nlargest(3, '2020')['Country'].tolist()
print(top_countries)
```

```python
iran_mean = np.float64(df[df['Country'] == 'Iran (Islamic Republic of)'].iloc[:, 1:].values.mean())
print(iran_mean)
```

```python
highest_growth = df.loc[(df['2020'] - df['1990']).idxmax(), 'Country']
print(highest_growth)
```
